### GRAFICO DE LINEAS

```python
# Usaremos el mismo fichero de datos_pedidos

# Vamos a ver un grafico de lineas para ver la evolucion del importe por mes
# 0. Importar las librerias necesarias
import pandas as pd
import matplotlib.pyplot as plt

# 1. Ajustar la ruta del fichero excel

ruta_excel = "/Users/rogerdefez/Documents/Cursos i Llibres/Curso Barcelona Activa/2026-04-29_Python3_Ciencia-de-datos-Python-Institute-PCAD-31-Ed2/Datos_originales//Datos Pedidos.xlsx"

# 2. Cargar el DataFrame excel con los datos
pedidos = pd.read_excel(ruta_excel, sheet_name="Datos")

# 3. Crear una columna que contenga el mes en letras (Es) partiendo de la columna fecha de pedido

pedidos['Mes'] = pedidos['Fecha de Pedido'].dt.month_name(locale='es_ES.UTF-8')

# 4. Agrupar por mes y sumar importes (despreciamos o depreciamos el año)

importe_meses = pedidos.groupby('Mes')['Importe'].sum().reset_index()

# importe_meses = pedidos.groupby('Mes').Importe.sum()   ## Es lo mismo sin corchete si nombre no tiene espacios en blanco
# type(importe_meses).  vemos que tipo de objeto es
# con reset_index() al final convierte la serie en un dataframe

# 5. Como ordenar los meses cronologicamente, primero vamos a crear una lista

orden_meses = ['Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo', 'Junio',
               'Julio', 'Agosto', 'Septiembre', 'Octubre', 'Noviembre', 'Diciembre']

# 6. Usar el metodo Categorical de pandas para ordenar según la lista

importe_meses['Mes'] = pd.Categorical(importe_meses['Mes'], categories=orden_meses, ordered=True)

importe_meses


# 7. Ordenar el dataframe

importe_meses = importe_meses.sort_values(by='Mes')
importe_meses = importe_meses.reset_index(drop=True)

# 8. VAmos a dibujar Crear una instancia de matplotlib con ax

fig, ax=plt.subplots(figsize=(10,4))
ax.plot(importe_meses['Mes'],importe_meses['Importe'],marker='o', color='red',markerfacecolor='green')
ax.set_xticklabels(importe_meses['Mes'],rotation=45)
plt.show

```
