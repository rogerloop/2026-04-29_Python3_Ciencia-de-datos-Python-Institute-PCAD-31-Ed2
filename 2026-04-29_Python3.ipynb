Practica 2 : Mediante el fichero de excel "DAtos telefonia-..." consolidar los meses de enero a junio en un solo dataframe, crear una columna que me de el "importe factura neto" que sera igual al importe fra despues de aplicar un descuento del 8,39% solo a los clientes que necieron los dias 1, 5 o 10 de cualquier mes. Para el resto el Importe Fraa neto = Importe FRA 
Queremos tener en una lista las diferentes localidades del campo localidad y despues queremos ver la suma de importe FRa neto de los clientes del servicio ADSL + Movil (dos servicios) 

```python
# Importar y consolidar datos Limpiand nulos si los hay

fras_febrero = pd.read_excel(ruta_fichero,sheet_name="Facturación Febrero 2020")
fras_febrero = fras_febrero.dropna(how='all')

fras_marzo = pd.read_excel(ruta_fichero,sheet_name="Facturación Marzo 2020",header=1)
fras_marzo = fras_marzo.dropna(how='all')

fras_abril = pd.read_excel(ruta_fichero,sheet_name="Facturación Abril 2020",header=1)
fras_abril = fras_abril.dropna(how='all')

fras_mayo = pd.read_excel(ruta_fichero,sheet_name="Facturación Mayo 2020",header=1)
fras_mayo = fras_mayo.dropna(how='all')

fras_junio = pd.read_excel(ruta_fichero,sheet_name="Facturación Junio 2020",header=2)
fras_junio = fras_junio.dropna(how='all')

fras_feb_jun = pd.concat([fras_febrero, fras_marzo, fras_abril, fras_mayo, fras_junio])
fras_feb_jun['Mes'] = fras_feb_jun['Fecha Nacimiento'].dt.month

fras_semestre = pd.concat([fras_enero, fras_feb_jun])
fras_semestre_AN1510 =  fras_semestre[(fras_semestre['Fecha Nacimiento'].dt.day == 1) | 
                        (fras_semestre['Fecha Nacimiento'].dt.day == 5) | 
                        (fras_semestre['Fecha Nacimiento'].dt.day == 10)]
fras_semestre_AN1510['importe factura neto'] = round(fras_semestre_AN1510['Importe factura']*(1-0.0839),2)

fras_semestre_ANNO1510 =  fras_semestre[(fras_semestre['Fecha Nacimiento'].dt.day != 1) | 
                        (fras_semestre['Fecha Nacimiento'].dt.day != 5) | 
                        (fras_semestre['Fecha Nacimiento'].dt.day != 10)]
fras_semestre_ANNO1510['importe factura neto'] = 0
fras_semestre = pd.concat([fras_semestre_AN1510, fras_semestre_ANNO1510])

localidad_fras_semestre = fras_semestre['Localidad'].unique
print(localidad_fras_semestre)

filtro_ADSL_Movil = (fras_semestre['Servicios'] == 'ADSL') | (fras_semestre['Servicios'] == 'Móvil')
suma_importe_FRA_neto_filtradas = fras_semestre[filtro_ADSL_Movil]['importe factura neto'].sum()

print(f"La suma de los importes factura neto de los clientes de servicios ADSL y Móvil es: {suma_importe_FRA_neto_filtradas}")
```
