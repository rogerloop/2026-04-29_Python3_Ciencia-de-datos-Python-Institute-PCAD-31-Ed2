Los diferentes tipos de variables en tkinter  
- StringVar()  = Tipo str y sirve para texto (label)  
- IntVar() = Tipo int y sirve para enteros, check button, radio button
- DoubleVar() = Tipo float y sirve para decimales
- BooleanVar() = Tipo bool y sirve para True o False 

Practica 7: Partiendo del csv Importe Franjas, poner en un formulario de tkinter centrado, una etiqueta con el Importe total de las facturas, añadir un botón que me aplique un incremento del 15% al Importe total y me lo muestre en la etiqueta

```python
#Solución Practica 7
#0. Importar librerias
import tkinter as tk
import pandas as pd

#1. Ajustar la ruta del .csv
ruta_csv = r"C:\PCAD\Importe Franjas.csv"

#2. Cargar el dataframe con los datos de los importes por franja
facturas = pd.read_csv(ruta_csv)

#3. Guardar en una referencia la suma del importe
importe_neto = facturas['Importe factura neto'].sum()

#4. Crear la ventana principal
root = tk.Tk()
root.title("Analisis Franjas")

#5. Guardar dimensiones en variables
ancho_ventana = 500
alto_ventana = 300
ancho_pantalla = root.winfo_screenwidth()
alto_pantalla = root.winfo_screenheight()

#6. Calcular la posicion de la izquierda (x) y de arriba (y)
pos_x =(ancho_pantalla // 2) - (ancho_ventana // 2)
pos_y =(alto_pantalla // 2) - (alto_ventana // 2)

#7. Geometry root.geometry("500x300+100+50")
root.geometry(f"{ancho_ventana}x{alto_ventana}+{pos_x}+{pos_y}")
print("El importe es",importe_neto)

#8. Crear el frame o contenedor para ordenar los widgets
frm = tk.LabelFrame(root, text="Métricas", padx=5, pady=5)
frm.pack(padx=5, pady=5, fill="x")

#8.1 Crear una función para formatear el numero raw de python 898999.36
def formato_euro(numero):
    #Que me ponga una coma para los miles si viene 590008.53 = 590,008.53
    txt = f"{numero:,.2f}"
    #Convertir a formato euro
    txt = txt.replace(",","x").replace(".",",").replace("x",'.')
    #Devolver el valor añadiendo el euro €
    return txt + " €"

#9. Crear variables para mostrar en pantalla de texto
vimporte = tk.StringVar(value=str(formato_euro(importe_neto)))

#10. Mostrar el importe factura neto en pantalla
tk.Label(frm, text="Importe neto:").grid(row=2, column=1, sticky="w", pady=4)
tk.Label(frm, textvariable=vimporte).grid(row=2, column=2, sticky="w", pady=4)

#10.1 Vamos a crear la función para incrementar un 15% el importe neto
#Crear un botón para actualizar un dato
def modificar_importe():
    resultado = importe_neto * 1.15
    return vimporte.set(formato_euro(resultado))

#10.2 Crear el botón para incrementar el importe en un 15%
tk.Button(root, text="Incrementar 15% importe", command=modificar_importe).pack(pady=5)

#10.3 Crear una caja de texto o entry, sirve para que el usuario pueda escribir un valor
entry = tk.Entry(root)
entry.pack(pady=5)

#11. Mostrar la ventana
root.mainloop()
```

```python
#Solución Practica 8
#0. Importar librerias
import tkinter as tk
import pandas as pd

#1. Ajustar la ruta del .csv
ruta_csv = r"C:\PCAD\Importe Franjas.csv"

#2. Cargar el dataframe con los datos de los importes por franja
facturas = pd.read_csv(ruta_csv)

#3. Guardar en una referencia la suma del importe
importe_neto = facturas['Importe factura neto'].sum()

#4. Crear la ventana principal
root = tk.Tk()
root.title("Analisis Franjas")

#5. Guardar dimensiones en variables
ancho_ventana = 500
alto_ventana = 300
ancho_pantalla = root.winfo_screenwidth()
alto_pantalla = root.winfo_screenheight()

#6. Calcular la posicion de la izquierda (x) y de arriba (y)
pos_x =(ancho_pantalla // 2) - (ancho_ventana // 2)
pos_y =(alto_pantalla // 2) - (alto_ventana // 2)

#7. Geometry root.geometry("500x300+100+50")
root.geometry(f"{ancho_ventana}x{alto_ventana}+{pos_x}+{pos_y}")
print("El importe es",importe_neto)

#8. Crear el frame o contenedor para ordenar los widgets
frm = tk.LabelFrame(root, text="Métricas", padx=5, pady=5)
frm.pack(padx=5, pady=5, fill="x")

#8.1 Crear una función para formatear el numero raw de python 898999.36
def formato_euro(numero):
    #Que me ponga una coma para los miles si viene 590008.53 = 590,008.53
    txt = f"{numero:,.2f}"
    #Convertir a formato euro
    txt = txt.replace(",","x").replace(".",",").replace("x",'.')
    #Devolver el valor añadiendo el euro €
    return txt + " €"

#9. Crear variables para mostrar en pantalla de texto
vimporte = tk.StringVar(value=str(formato_euro(importe_neto)))
vvalidacion = tk.StringVar(value=str("Pendiente"))

#10. Mostrar el importe factura neto en pantalla
tk.Label(frm, text="Importe neto:").grid(row=2, column=1, sticky="w", pady=4)
tk.Label(frm, textvariable=vimporte).grid(row=2, column=2, sticky="w", pady=4)

tk.Label(frm, text="Validación %:").grid(row=4, column=1, sticky="w", pady=4)
tk.Label(frm, textvariable=vvalidacion).grid(row=4, column=2, sticky="w", pady=4)

#10.1 Función que va a validar si el valor de entry o de la caja de texto es decimal o float, para ello usaremo try...expect, para interceptar un error y darle tratamiento, except hay muchos
def validar_decimal(x):
    try:
        numero = float(x)
        vvalidacion.set("% válido")
        return numero
    except ValueError:
        vvalidacion.set("% no válido")

#10.2 Vamos a crear la función para incrementar un 15% el importe neto
#Crear un botón para actualizar un dato
def modificar_importe():
    porcentaje = validar_decimal(entry.get())
    if vvalidacion.get() == "% válido":
        resultado = importe_neto * (1 + (porcentaje/100)) 
        return vimporte.set(formato_euro(resultado))

#10.3 Crear el botón para incrementar el importe en un 15%
tk.Button(root, text="Incrementar importe", command=modificar_importe).pack(pady=5)

#10.3 Crear una caja de texto o entry, sirve para que el usuario pueda escribir un valor
entry = tk.Entry(root)
entry.pack(pady=5)

#11. Mostrar la ventana
root.mainloop()
```

Queremos visualizar el dataframe dentro de un formulario de tkinter

```python
#Solución Practica 8
#0. Importar librerias
import tkinter as tk
from tkinter import ttk
import pandas as pd

#1. Ajustar la ruta del .csv
ruta_csv = r"C:\PCAD\Importe Franjas.csv"

#2. Cargar el dataframe con los datos de los importes por franja
facturas = pd.read_csv(ruta_csv, index_col=0)

#4. Crear la ventana principal
root = tk.Tk()
root.title("Analisis Franjas")

#5. Guardar dimensiones en variables
ancho_ventana = 500
alto_ventana = 300
ancho_pantalla = root.winfo_screenwidth()
alto_pantalla = root.winfo_screenheight()

#6. Calcular la posicion de la izquierda (x) y de arriba (y)
pos_x =(ancho_pantalla // 2) - (ancho_ventana // 2)
pos_y =(alto_pantalla // 2) - (alto_ventana // 2)

#7. Geometry root.geometry("500x300+100+50")
root.geometry(f"{ancho_ventana}x{alto_ventana}+{pos_x}+{pos_y}")
print("El importe es",importe_neto)

#8. Crear el frame o contenedor para ordenar los widgets
frm = tk.LabelFrame(root, text="Métricas", padx=5, pady=5)
frm.pack(padx=5, pady=5, fill="x")

#9. Crear un widget Treeview (lo tenemos dentro de la clase ttk, la tendremos que llamar primero)
tabla = ttk.Treeview(root)   #Widget treeview
tabla.pack(fill="both", expand=True)

#10. Definir columnas, con la lista de nombres del dataframe
tabla['columns'] = list(facturas.columns)
tabla['show'] = "headings"

#11. Cabeceras
for columna in facturas.columns:
    tabla.heading(columna, text=columna)
    tabla.column(columna, width=120)

#12. Insertar las filas
for _, fila in facturas.iterrows():
    tabla.insert("","end",values=list(fila))

#13. Mostrar la ventana
root.mainloop()
```
