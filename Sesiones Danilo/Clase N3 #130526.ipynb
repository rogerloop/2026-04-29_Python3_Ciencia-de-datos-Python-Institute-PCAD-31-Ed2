```python
#Solución de la practica 11
#Sentencia SQL para exportar los datos al CSV
"""SELECT
	TOP(10) 
	P.PaísDestinatario AS Pais,
	SUM(DP.Cantidad * DP.PrecioUnidad) AS Importe
FROM
	Pedidos P, [Detalles de pedidos] DP
WHERE
	P.IdPedido = DP.IdPedido
GROUP BY
	P.PaísDestinatario
ORDER BY
	Importe DESC;"""

#0. Llamar a las librerias necesarias
import pandas as pd
import matplotlib.pyplot as plt

#1. Ajustar la ruta del fichero .csv
ruta_fichero = r"C:\PCAD\Importe Pais.csv"

#2. Cargar los datos del csv en un DataFrame
importe_pais = pd.read_csv(ruta_fichero, sep=';', header=None, names=['Pais','Importe'])

#3. Comprobar los datos
importe_pais.info()

#4. Cambiar la coma por el punto + convertir a float64
importe_pais['Importe'] = importe_pais['Importe'].str.replace(',','.').astype('float64')

#5. Comprobar los datos
print()
importe_pais.info()

#6. Dibujar el grafico de barras
plt.figure(figsize=(10,6))
importe_pais.plot(kind='bar',x='Pais',y='Importe',color='orange',label='Pais')
plt.title("Suma Importe por Pais")
plt.xticks(rotation=-75)
plt.show()
```
