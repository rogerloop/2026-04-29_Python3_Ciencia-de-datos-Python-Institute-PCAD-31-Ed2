Practica 10: Mediante el dataframe de pedidos agrupados, ver en un grafico circular los 3 paises que menos importe tienen, con sus % reales.
Si en el dataframe el % de un pais es el 15,38%, en el grafico tengo que ver
ese mismo porcentaje.
Pista!!! Si queremos los 3 que menos, necesitaremos 4 porciones.

```python
#0. Importar las librerias necesarias
import pandas as pd

#1. Ajustar la ruta del excel
ruta_excel = r"C:\PCAD\Datos Pedidos.xlsx"

#2. Cargar el dataframe con los datos
pedidos = pd.read_excel(ruta_excel, sheet_name="Datos")

#3. Crear la agrupación por Pais + la suma de Importe
pedidos.groupby(['Pais'])['Importe'].sum()

#Que nos devuelve este codigo, devuelve una serie
type(pedidos.groupby(['Pais'])['Importe'].sum())

#Como obtener varias metricas de una dimension, con agg
pedidos.groupby(['Pais'])['Importe'].agg(['sum','mean','count'])

#Como convertir esta agrupación en un dataframe correcto, con reset_index al final
pedidos.groupby(['Pais'])['Importe'].agg(['sum','mean','count']).reset_index()

#Trabajar con una referencia a este dataframe
pedidos_agrupados = pedidos.groupby(['Pais'])['Importe'].agg(['sum','mean','count']).reset_index()
```

```python
#Solución Practica 10
#0. Comprobar que puedo leer el dataframe pedidos_agrupados
pedidos_agrupados
#1. Crear una columna con el % de la sum con respecto al total
#almacenar en una variable la suma total
suma_total = round(pedidos_agrupados['sum'].sum(),2)
print('La suma total es',suma_total)
#2. Crear una columna con el %
pedidos_agrupados['% sum'] = round((pedidos_agrupados['sum'] / suma_total) * 100,2)
#3. Mostrar el dataframe en orden descendente para ver los ultimos %
pedidos_agrupados.sort_values(by='% sum', ascending=True)
#4. Que datos deberia ver en el grafico
#   Polonia - 0,25%
#   Noruega - 0,42%
#   Argentina - 0,60%
#5. Guardar en un dataframe solo los 3 que menos (nlargest = los que +, nsmallest los que menos)
pedidos_agrupados_3Men = pedidos_agrupados.nsmallest(3, "sum")
#6. Guardar en un dataframe el resto de paises
pedidos_agrupados_resto = pedidos_agrupados.nlargest(len(pedidos_agrupados)-3,"sum")
#7. Guardar en un referencia la suma del resto
suma_resto = pedidos_agrupados_resto['sum'].sum()
#8. Crear un dataframe con el pais = resto paises y el sum con la variable suma_resto
pedidos_grafico = pd.DataFrame({
    "Pais":["Resto paises"],
    "sum":[suma_resto]
})
#9. Concatenar el dataframe con los 3 que menos + el que contiene la suma del resto
pedidos_grafico = pd.concat([pedidos_agrupados_3Men,pedidos_grafico],ignore_index=True)
#10. Vamos con el grafico
import matplotlib.pyplot as plt
#11. Reservar la figure (cuadrado)
plt.figure(figsize=(6,6))
plt.pie(pedidos_grafico['sum'],labels=pedidos_grafico['Pais'],autopct='%1.2f%%')
plt.legend(title="Paises", bbox_to_anchor=(1.05,1))   #la caja de la leyenda esquina superior derecha
plt.show()
```

```python
#Grafico de lineas, es el grafico por defecto de matplotlib kind='line'
#Usaremos el mismo fichero de Datos pedidos.xlsx
#Es un grafico de lineas para ver la evolución del Importe por mes
#0. Importar las 2 librerias necesarias
import pandas as pd
import matplotlib.pyplot as plt

#1. Ajustar la ruta del excel
ruta_excel = r"C:\PCAD\Datos Pedidos.xlsx"

#2. Cargar el dataframe con los datos
pedidos = pd.read_excel(ruta_excel, sheet_name="Datos")

#3. Crear una columna que contenga el mes en letras (Es) partiendo de la columa "Fecha de pedido"
pedidos['Mes'] = pedidos['Fecha de Pedido'].dt.month_name(locale='es_ES.UTF-8')

#4. Agrupar por Mes y la agregación la suma de importe (depreciamos el año)
importe_meses = pedidos.groupby('Mes')['Importe'].sum().reset_index()
#importe_meses = pedidos.groupby('Mes').Importe.sum()   Es lo mismo pero sin []
#con reset_index() al final convierte la serie en un dataframe
#type(importe_meses)    #con type vemos que tipo de objeto es

#5. Como ordenar los meses cronologicamente, primero vamos a crear una lista
orden_meses = ['Enero','Febrero','Marzo','Abril','Mayo','Junio',
               'Julio','Agosto','Septiembre','Octubre','Noviembre','Diciembre']

#6. Usar el metodo Categorical de pandas para ordenar según la lista
importe_meses['Mes'] = pd.Categorical(importe_meses['Mes'], categories=orden_meses, ordered=True)

#7. Ordenar el dataframe
importe_meses = importe_meses.sort_values(by='Mes')
importe_meses = importe_meses.reset_index(drop=True)

#8. Crear una instancia de matplotlib con ax (puede ser cualquier nombre)
fig, ax=plt.subplots(figsize=(10,4))    #reservar espacio en el kernel
ax.plot(importe_meses['Mes'],importe_meses['Importe'],marker='o',color='red',markerfacecolor='green')
ax.set_xticklabels(importe_meses['Mes'], rotation=45)
plt.show()
```
