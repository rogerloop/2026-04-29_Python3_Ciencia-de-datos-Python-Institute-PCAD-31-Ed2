Practica 2: Mediante el fichero de excel "Datos telefonia separ...", consolidar los meses de Enero a Junio en un solo
dataframe, crear una columna que me de el "importe factura neto" = Importe factura despues de aplicar un descuento del 8,39% solo
a los clientes que nacieron los dias 1, 5 o 10 del mes correspondiente. Importe factura neto = Importe factura
Queremos tener en una lista las diferentes localidades del campo Localidad
Queremos ver la suma de importe factura neto de los clientes del
servicio ADSL + Movil

```python
import pandas as pd
ruta_fichero = r"C:\PCAD\Datos Telefonia Separados Meses Comerciales URL.xlsx"
fras_enero = pd.read_excel(ruta_fichero,sheet_name="Facturación Enero 2020", header=2)
fras_febrero = pd.read_excel(ruta_fichero,sheet_name="Facturación Febrero 2020", header=0)
fras_marzo = pd.read_excel(ruta_fichero,sheet_name="Facturación Marzo 2020", header=1)
fras_abril = pd.read_excel(ruta_fichero,sheet_name="Facturación Abril 2020", header=1)
fras_mayo = pd.read_excel(ruta_fichero,sheet_name="Facturación Mayo 2020", header=1)
fras_junio = pd.read_excel(ruta_fichero,sheet_name="Facturación Junio 2020", header=2)
#Voy a comprobar un DataFrame
fras_enero
#Concatenar los 6 meses
fras_semestre = pd.concat([fras_enero,fras_febrero,fras_marzo,fras_abril, fras_mayo, fras_junio],ignore_index=True)
#Comprobar los tipos de datos
#fras_semestre.info()
fras_semestre['Dia Nac'] = fras_semestre['Fecha Nacimiento'].dt.day.astype('Int64')  #Con astype convierto el tipo de dato
fras_semestre
#Eliminar las filas en blanco
fras_semestre = fras_semestre.dropna(how='all')
#Opción 1 con el or |
#(fras_semestre['Dia Nac']==1) | (fras_semestre['Dia Nac']==5) |
#(fras_semestre['Dia Nac']==1)
#Opción 2 con el isin dentro de una lista
fras_clientes_1 = fras_semestre[fras_semestre['Dia Nac'].isin([1,5,10])]
#Comprobar si los numeros unicos coinciden
fras_clientes_1['Dia Nac'].unique()
#En un dataframe el resto de clientes que no nacieron ese dia
#la virgulilla hace lo contrario a lo que viene a continuación
#alt + 126 o AltGr + 4
fras_clientes_2 = fras_semestre[~fras_semestre['Dia Nac'].isin([1,5,10])]
fras_clientes_2
fras_clientes_2['Dia Nac'].unique()
#Vamos a calcular el neto para clientes 1 (1, 5, 10)
fras_clientes_1['Importe factura neto'] = fras_clientes_1['Importe factura'] * (1-0.0839)
fras_clientes_2['Importe factura neto'] = fras_clientes_2['Importe factura']
#Concatenar los 2 dataframes (reasigno el mismo nombre)
fras_semestre = pd.concat([fras_clientes_1,fras_clientes_2])
#Filtro por sericio
filtro_servicio = (fras_semestre['Servicios']=='Móvil') |(fras_semestre['Servicios']=='ADSL')
filtro_importe_neto = (fras_semestre['Importe factura neto']>30)
fras_semestre = fras_semestre[filtro_servicio & filtro_importe_neto]
suma_importe = round(fras_semestre['Importe factura neto'].sum(),2)
print("El valor del importe en ADSL + Móvil es",suma_importe)
```

```python
#Es calcular la edad actual en función de una fecha de nacimiento
import pandas as pd
alumnos = {'Nombre':['Maria','Luis','Alvaro','Sonia'],
           'Fecha Nac':['1994-05-14','1985-06-20','2001-03-07','1978-11-18']}
clase = pd.DataFrame(data=alumnos)
clase
#Que la columna no es fecha, toca convertir
clase['Fecha Nac'] = pd.to_datetime(clase['Fecha Nac'])
#Ver año, mes y dia
clase['Fecha Nac'].dt.year
clase['Fecha Nac'].dt.month
clase['Fecha Nac'].dt.day
clase['Fecha Nac'].dt.dayofweek      #Dia semana del 1 al 7
clase['Fecha Nac'].dt.day_name()     #Dia de la semana en letras
clase['Fecha Nac'].dt.month_name()   #Mes en letras
#Guardar en una columna el numero de dias entre la Fecha Nac y hoy, un metodo de pandas que equivale a timedelta, se llama timestamp
clase['Dias'] = (pd.Timestamp('today') - clase['Fecha Nac']).dt.days
#calcular la edad aproximada, solo cogemos la parte entera del resultado
clase['Edad'] = clase['Dias'] // 365
#Calcular su decada teniendo que va de 
# 10-19 = Decada 10
# 20-29 = Decada 20
# 30-39 = Decada 30
clase['Decada'] = (clase['Edad'].astype('str').str[0]+'0').astype('Int64')
clase
```

Practica 3: Partiendo del DataFrame 'fras_semestres', averiguar la suma de importe neto de los clientes que nacieron un mes de mayo o septiembre. (Necesitaremos averiguar primero el mes)
Calcular la edad del cliente y si es mayor de 40 años, poner en la columna nueva 'VIP' = SI, si no VIP = NO
Crear una columna que nos diga 15 dias antes en que dia y mes debemos enviar al cliente una felicitacion de cumpleaños

```python
#Practica 3.1 Partiendo del DataFrame 'fras_semestres', averiguar la suma de importe neto de los clientes que nacieron un mes de mayo o septiembre. (Necesitaremos averiguar primero el mes)
#Crear una columna que contenga el mes de nacimiento
fras_semestre['Mes Nac'] = fras_semestre['Fecha Nacimiento'].dt.month
#Voy a filtrar y sumar en la misma linea
resultado = fras_semestre[fras_semestre['Mes Nac'].isin([5,9])]['Importe factura neto'].sum()
print("El resultado es",round(resultado,2))
```

```python
#Practica 3.2 Calcular la edad del cliente y si es mayor de 40 años, poner en la columna nueva 'VIP' = SI, si no VIP = NO
fras_semestre['Dias'] = (pd.Timestamp('today') - fras_semestre['Fecha Nacimiento']).dt.days
fras_semestre['Edad'] = fras_semestre['Dias'] // 365
#Voy a crear una función para aplicar al nuevo campo
def clasificar_vip(edad):
    if edad>40:
        return "SI VIP"
    else:
        return "NO VIP"
#Crear el campo VIP
fras_semestre['VIP'] = fras_semestre['Edad'].apply(clasificar_vip)
fras_semestre
```

```python
#Practica 3.3 Crear una columna que nos diga 15 dias antes en que dia y mes debemos enviar al cliente una felicitacion de cumpleaños
fras_semestre['Fecha Aviso'] = fras_semestre['Fecha Nacimiento'] - timedelta(days=15)
fras_semestre['Fecha Aviso'] = fras_semestre['Fecha Aviso'].dt.strftime('%d/%m')
fras_semestre
```
