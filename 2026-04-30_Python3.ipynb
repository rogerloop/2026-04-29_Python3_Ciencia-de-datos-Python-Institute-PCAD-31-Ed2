```python
# Es calcular la edat actual en funcion de una fecha de nacimiento
import pandas as pd

alumnos = {'Nombre' : ['Maria', 'Luis', 'Alvaro', 'Sonia'],
           'Fecha Nac': ['1994-05-14','1985-06-20','2001-03-07','1978-11-18']}
clase = pd.DataFrame(data=alumnos)

# LA columna no es fecha
clase['Fecha Nac'] = pd.to_datetime(clase['Fecha Nac'])

# Ver año mes y dia
clase['Fecha Nac'].dt.year
clase['Fecha Nac'].dt.month
clase['Fecha Nac'].dt.day
clase['Fecha Nac'].dt.dayofweek #dia semana del 1 al 7
clase['Fecha Nac'].dt.day_name() #Dia semana en letras
clase['Fecha Nac'].dt.month_name() 

#Guardar en una columna el numero de dias entre la fecha de nacimiento y hoy
# es un metodo de pandas que equivale a timedelta, se llama timestamp

clase['Dias'] = (pd.Timestamp('today') - clase['Fecha Nac']).dt.days

#calcular la edasd aproximada
clase['Edad'] = clase['Dias'] // 365

#calcular su decada teniendo en cuenta que va de 
# 10-19 = Decada 10
# 20-29 = Decada 20
# 30-39 = Decada 30
clase['Decada'] = (clase['Edad'].astype('str').str[0] + '0').astype('Int64')
clase
```
