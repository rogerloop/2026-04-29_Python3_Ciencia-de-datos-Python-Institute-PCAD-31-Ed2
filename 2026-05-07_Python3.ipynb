```python
import pandas as pd
import tkinter as tk

importe_franjas = pd.read_csv("Importe franjas.csv")

importe_neto = importe_franjas['Importe factura neto'].sum()



# 1. Crear la ventana principal
root = tk.Tk()

root.title("Analisis Franjas")

# 2. Vamos a ajustar las medidas de la ventana. Guardar dimensiones en variables

ancho_ventana = 500
alto_ventana = 300
ancho_pantalla =root.winfo_screenwidth()
alto_pantalla =root.winfo_screenheight()

# Calcular la posicion de la izquierda (x) y de arriba (y)

pos_x = (ancho_pantalla // 2) - (ancho_ventana // 2)
pos_y = (alto_pantalla // 2) - (alto_ventana // 2)

# root.geometry("500x300+100+50")

root.geometry(f"{ancho_ventana}x{alto_ventana}+{pos_x}+{pos_y}")

# 8. Vamos a crear el frame o contenedor para ordenar los windgets

frm = tk.LabelFrame(root, text="Métricas", padx=5, pady=5)
frm.pack(padx=5, pady=5, fill="x" )

# 8.1 Crear una funcion para dar formato al numero raw de python 9090998989.36

def formato_euro(numero):
    #Que me pònga una coma para los miles si viene 590008.83 = 590,008.53
    txt = f"{numero:,.2f}"

    #Convertir a formato euro
    txt = txt.replace(",","x").replace(".",",").replace("x",".")
    # DEvolver valor añadiendo el €
    return txt + " €"


# 9. Crear variables para mostrar en pantalla

vimporte = tk.StringVar(value=str(formato_euro(importe_neto)))

# 10. Mostrar en pantalla

tk.Label(frm,text="Importe neto:").grid(row=2, column=1, sticky="w", pady=4,)
tk.Label(frm,textvariable=vimporte).grid(row=2, column=2, sticky="w", pady=4)


# 10.1 Vamos a crear la funcion para incrementar el importe 15%

def modificar_importe():
    resultado = importe_neto * 1.15
    return vimporte.set(formato_euro(resultado))

# 10.2 Crear un boton para actualizar un dato que se muestra en pantalla

tk.Button(root, text="Incremento Importe 15%", command=modificar_importe).pack(pady=5)


# 10.3 Crear una caja que permite escribir o introducir caracteres

entry = tk.Entry(root)
entry.pack(pady=5)

# 3. Bucle principal mantiene la ventana viva y también escucha eventos

root.mainloop()
```

PRACTICA 8 : Copiando la celda anterior , pondremos una caja de texto, donde poner el valor del incremento ,  al hacer clic en el boton de modificar importe ver el nuevo resultado.
para capturar el valor de entry tengo que hacer entry.get()

```python
import pandas as pd
import tkinter as tk

importe_franjas = pd.read_csv("Importe franjas.csv")

importe_neto = importe_franjas['Importe factura neto'].sum()



# 1. Crear la ventana principal
root = tk.Tk()

root.title("Analisis Franjas")

# 2. Vamos a ajustar las medidas de la ventana. Guardar dimensiones en variables

ancho_ventana = 500
alto_ventana = 300
ancho_pantalla =root.winfo_screenwidth()
alto_pantalla =root.winfo_screenheight()

# Calcular la posicion de la izquierda (x) y de arriba (y)

pos_x = (ancho_pantalla // 2) - (ancho_ventana // 2)
pos_y = (alto_pantalla // 2) - (alto_ventana // 2)

# root.geometry("500x300+100+50")

root.geometry(f"{ancho_ventana}x{alto_ventana}+{pos_x}+{pos_y}")

# 8. Vamos a crear el frame o contenedor para ordenar los windgets

frm = tk.LabelFrame(root, text="Métricas", padx=5, pady=5)
frm.pack(padx=5, pady=5, fill="x" )

# 8.1 Crear una funcion para dar formato al numero raw de python 9090998989.36

def formato_euro(numero):
    #Que me pònga una coma para los miles si viene 590008.83 = 590,008.53
    txt = f"{numero:,.2f}"

    #Convertir a formato euro
    txt = txt.replace(",","x").replace(".",",").replace("x",".")
    # DEvolver valor añadiendo el €
    return txt + " €"


# 9. Crear variables para mostrar en pantalla

vimporte = tk.StringVar(value=str(formato_euro(importe_neto)))
vvalidacion = tk.StringVar(value=str("Pendiente"))

# 10. Mostrar en pantalla

tk.Label(frm,text="Importe neto:").grid(row=2, column=1, sticky="w", pady=4,)
tk.Label(frm,textvariable=vimporte).grid(row=2, column=2, sticky="w", pady=4)

tk.Label(frm,text="Validación %").grid(row=3, column=1, sticky="w", pady=4,)
tk.Label(frm,textvariable=vvalidacion).grid(row=3, column=2, sticky="w", pady=4)



# 10.1 Funcion que va a validar si el valor de entry  o de la caja de texto es decimal o float, para ello usaremo stry...except
def validar_decimal(x):
    try:
        numero = float(x)
        vvalidacion.set("% válido")
        return numero

    except ValueError:
        vvalidacion.set("% no válido")

# 10.2 Vamos a crear la funcion para incrementar el importe 15%

entry = tk.Entry(root)
entry.pack(pady=5)

def modificar_importe():
    porcentaje = validar_decimal(entry.get())
    if vvalidacion.get() == "% válido":
        resultado = importe_neto * (1 + (porcentaje / 100))
    return vimporte.set(formato_euro(resultado))
    

# 10.2 Crear un boton para actualizar un dato que se muestra en pantalla

tk.Button(root, text="Incremento importe", command=modificar_importe).pack(pady=5)






# 3. Bucle principal mantiene la ventana viva y también escucha eventos

root.mainloop()
```

Queremos  visualizar el dataframe dentro de un formiulario de tkinter

```python
import pandas as pd
import tkinter as tk
from tkinter import ttk

facturas = pd.read_csv("Importe franjas.csv")

# 1. Crear la ventana principal
root = tk.Tk()

root.title("Analisis Franjas")

# 2. Vamos a ajustar las medidas de la ventana. Guardar dimensiones en variables

ancho_ventana = 500
alto_ventana = 300
ancho_pantalla =root.winfo_screenwidth()
alto_pantalla =root.winfo_screenheight()

# Calcular la posicion de la izquierda (x) y de arriba (y)

pos_x = (ancho_pantalla // 2) - (ancho_ventana // 2)
pos_y = (alto_pantalla // 2) - (alto_ventana // 2)

# root.geometry("500x300+100+50")

root.geometry(f"{ancho_ventana}x{alto_ventana}+{pos_x}+{pos_y}")

# 8. Vamos a crear el frame o contenedor para ordenar los windgets

frm = tk.LabelFrame(root, text="Métricas", padx=5, pady=5)
frm.pack(padx=5, pady=5, fill="x" )


# 9. Crear un widget treeview. lo tenemos dentro de la clase ttk, la tendremos que llamar primero

tabla = ttk.Treeview(root)
tabla.pack(fill="both", expand=True)

#10 Definir colummnas con la lista de nombres del dataframe

tabla['columns'] = list(facturas.columns)
tabla['show'] = "headings"

# 11. Cabeceras 

for columna in facturas.columns:
    tabla.heading(columna, text=columna)
    tabla.column(columna, width=120)

# 12. Insertar las filas

for _, fila in facturas.iterrows():
    tabla.insert("","end",values=list(fila))


root.mainloop()
```
