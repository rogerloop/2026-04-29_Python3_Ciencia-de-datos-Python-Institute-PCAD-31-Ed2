```python
# crear un formulario y recrearlo en pantalla

# 0. Importar las librerias

import tkinter as tk

# 1. Crear la ventana principal
root = tk.Tk()

root.title("Formulario de prueba")

# 2. Vamos a ajustar las medidas de la ventana. Guardar dimensiones en variables

ancho_ventana = 500
alto_ventana = 300
ancho_pantalla =root.winfo_screenwidth()
alto_pantalla =root.winfo_screenheight()

# Calcular la posicion de la izquierda (x) y de arriba (y)

pos_x = (ancho_pantalla // 2) - (ancho_ventana // 2)
pos_y = (alto_pantalla // 2) - (alto_ventana // 2)

# root.geometry("500x300+100+50")

root.geometry(f"{ancho_ventana}x{alto_ventana}+{pos_x}+{pos_y}")

# 3. Bucle principal mantiene la ventana viva y también escucha eventos

root.mainloop()
```

```python
# crear un formulario y centrarlo en pantalla , mostrar un valor dentro del formulario
# Creamos un contenedor frame para poner los widgets

# 0. Importar las librerias

import tkinter as tk

# 1. Crear la ventana principal
root = tk.Tk()

root.title("Formulario de prueba")

# 2. Vamos a ajustar las medidas de la ventana. Guardar dimensiones en variables

ancho_ventana = 500
alto_ventana = 300
ancho_pantalla =root.winfo_screenwidth()
alto_pantalla =root.winfo_screenheight()

# Calcular la posicion de la izquierda (x) y de arriba (y)

pos_x = (ancho_pantalla // 2) - (ancho_ventana // 2)
pos_y = (alto_pantalla // 2) - (alto_ventana // 2)

# root.geometry("500x300+100+50")

root.geometry(f"{ancho_ventana}x{alto_ventana}+{pos_x}+{pos_y}")

# crear variables para mostrar en pantalla

vfilas = tk.StringVar(value=str(125))
vimporte = tk.StringVar(value=str(6895))

# Vamos a crear el frame o contenedor para ordenar los windgets

frm = tk.LabelFrame(root, text="Métricas", padx=5, pady=5)
frm.pack(padx=5, pady=5, fill="x" )


#mostrar en pantalla

tk.Label(frm,text="Filas:").grid(row=1, column=1, sticky="w", pady=4,)
tk.Label(frm,textvariable=vfilas).grid(row=1, column=2, sticky="w", pady=4)

tk.Label(frm,text="Importe:").grid(row=2, column=1, sticky="w", pady=4,)
tk.Label(frm,textvariable=vimporte).grid(row=2, column=2, sticky="w", pady=4)

# Crear un boton para actualizar un dato que se muestra en pantalla

def modificar_importe():
    return vimporte.set("15695")

tk.Button(root, text="Modificar importe", command=modificar_importe).pack(pady=5)





# 3. Bucle principal mantiene la ventana viva y también escucha eventos

root.mainloop()
```

Los diferente stipos de variables en tkinter
- StringVar() = Tipo str y sirvepara texto (POdemos usar en un label)
- IntVar() = Tipo int y sirve para enteros,, radioButton (Solo se puede elegir 1 es el redondito)       (ChekButton es el cuadradito permite seleccion multiple), 
- DoubleVar() = Tipo float y sirve para decimales
- BoleanVar() = Tipo bool y sirve para booleanos Verdadero Falso

PRACTICA 7: Partiendo del CSV importe franjas, poner en un formulario de tkinter centradso una etiquetacon el importe total de las facturas, añadir un boton que me aplique un incremento del 15% al importe total y me lo muestre en el label

```python
import pandas as pd
import tkinter as tk

importe_franjas = pd.read_csv("Importe franjas.csv")

total_facturas = importe_franjas['Importe factura neto'].sum()

total_facturas_15 = total_facturas * 1.15

# 1. Crear la ventana principal
root = tk.Tk()

root.title("Formulario de prueba")

# 2. Vamos a ajustar las medidas de la ventana. Guardar dimensiones en variables

ancho_ventana = 500
alto_ventana = 300
ancho_pantalla =root.winfo_screenwidth()
alto_pantalla =root.winfo_screenheight()

# Calcular la posicion de la izquierda (x) y de arriba (y)

pos_x = (ancho_pantalla // 2) - (ancho_ventana // 2)
pos_y = (alto_pantalla // 2) - (alto_ventana // 2)

# root.geometry("500x300+100+50")

root.geometry(f"{ancho_ventana}x{alto_ventana}+{pos_x}+{pos_y}")

# crear variables para mostrar en pantalla

vimporte = tk.StringVar(value=str(f"{round(total_facturas,2)} €" ))

# Vamos a crear el frame o contenedor para ordenar los windgets

frm = tk.LabelFrame(root, text="Métricas", padx=5, pady=5)
frm.pack(padx=5, pady=5, fill="x" )


#mostrar en pantalla

tk.Label(frm,text="Total Importe Facturas:").grid(row=2, column=1, sticky="w", pady=4,)
tk.Label(frm,textvariable=vimporte).grid(row=2, column=2, sticky="w", pady=4)

# Crear un boton para actualizar un dato que se muestra en pantalla

def modificar_importe():
    return vimporte.set(f"{round(total_facturas_15,2)} €")

tk.Button(root, text="Incremento 15%", command=modificar_importe).pack(pady=5)

# 3. Bucle principal mantiene la ventana viva y también escucha eventos

root.mainloop()
```
